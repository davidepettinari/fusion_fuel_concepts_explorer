```python
from pathlib import Path
import sys
from fusion_fuel_concepts_explorer.sources import SteadyStateSource

FONT_SIZE = 18

def style_source_plot(fig, ax, font_size=FONT_SIZE):
    """Remove the title and enlarge all text of a SteadyStateSource.plot() figure."""
    ax.set_title("")
    for a in fig.axes:
        a.xaxis.label.set_fontsize(font_size)
        a.yaxis.label.set_fontsize(font_size)
        a.tick_params(labelsize=font_size)
        a.yaxis.get_offset_text().set_fontsize(font_size)
    legend = ax.get_legend()
    if legend is not None:
        for text in legend.get_texts():
            text.set_fontsize(font_size)
    fig.tight_layout()
```

# DT

```python
# try density and temperature profiles
# Standard peaking profile:
# n(rho) = n_avg * p * (1-rho^2)^(p-1)
# This keeps <n>_rho = n_avg with 2*rho weighting and gives n(1)=0.
avg_density_DT = 1.3e20
avg_temperature_DT = 14.0
density_peaking_DT = 1.168
temperature_peaking_DT = 1.83
density_profile_DT = lambda rho: avg_density_DT * density_peaking_DT * ((1.0 - rho**2.0) ** (density_peaking_DT - 1.0))
temperature_profile_DT = lambda rho: avg_temperature_DT * temperature_peaking_DT * ((1.0 - rho**2.0) ** (temperature_peaking_DT - 1.0))
```

```python
DT_source = SteadyStateSource(
    source_type="DT",
    volume_m3=150.0,
    n_i=density_profile_DT,
    T_i=temperature_profile_DT,
    tau_p=2.5,
)
```

```python
print(DT_source)
print(DT_source.total_fusion_power_W/1e6, "MW")
fig, ax = DT_source.plot()
style_source_plot(fig, ax)
```
